```python
import os
import json
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
load_dotenv(override=True)
google_api_key = os.getenv('GOOGLE_API_KEY')
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=google_api_key)
```

```python
request = """
Tell me a joke about Indian democracy.
"""
messages = [{"role": "user", "content": request}]

completion = gemini.chat.completions.create(
    model="gemini-3.1-flash-lite",
    messages=messages,
    temperature=0.7,
)

display(Markdown(completion.choices[0].message.content))
```

```python
request = """
You are an newly joined associate at my office.
"""
messages = [{"role": "system", "content": request},
            {"role": "user", "content": "Hi My name is Amit"}
        ]

completion = gemini.chat.completions.create(
    model="gemini-3.1-flash-lite",
    messages=messages,
    temperature=0.7,
)

assistant_reply = completion.choices[0].message.content
display(Markdown(assistant_reply))
```
